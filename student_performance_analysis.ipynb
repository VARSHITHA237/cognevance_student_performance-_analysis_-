```python
import pandas as pd\nimport matplotlib.pyplot as plt\ndf = pd.read_csv('../data/student_performance.csv')\ndf.head()
```

## Data Cleaning

```python
df = df.drop_duplicates().dropna()\ndf.info()\ndf.describe()
```

## Exploratory Data Analysis

```python
subjects=['Mathematics','Science','English','History','Computer Science']\nprint(df[subjects].mean().sort_values(ascending=False))\nprint('Attendance/score correlation:', df['Attendance_Percentage'].corr(df['Overall_Score']))
```

```python
plt.figure(figsize=(8,5))\nplt.scatter(df['Attendance_Percentage'], df['Overall_Score'], alpha=.6)\nplt.xlabel('Attendance (%)'); plt.ylabel('Overall Score'); plt.title('Attendance vs Overall Score'); plt.show()
```

```python
df[subjects].mean().plot(kind='bar', figsize=(9,5), title='Average Performance by Subject'); plt.ylabel('Average Score'); plt.show()
```

## Key Insights\n- Higher attendance is positively associated with overall academic performance.\n- Students with stronger assignment averages and study time generally achieve higher scores.\n- Subject averages help identify areas where additional academic support may be useful.\n- Grade distribution can be used to identify students needing intervention.
